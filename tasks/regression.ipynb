```python
import numpy as np
import matplotlib.pyplot as plt
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([40, 48, 70, 78, 95])
```

### Initialize m and b with random/bad guesses


```python
m = 0.0   # slope guess
b = 0.0   # intercept guess
learning_rate = 0.01   # how big a step we take each time
epochs = 1000          # how many times we repeat the "nudge" process
 
n = len(hours)   # number of data points
```

### The gradient descent loop

```python
for epoch in range(epochs):
    predictions = m * hours + b                  # current line's guess for each point
    error = marks - predictions                  # residuals
 
    # These two formulas tell us WHICH DIRECTION to nudge m and b
    # (this is the "calculus part" we said not to worry about — just trust the direction)
    m_gradient = -(2/n) * np.sum(hours * error)
    b_gradient = -(2/n) * np.sum(error)
 
    # Take a small step in the direction that reduces error
    m = m - learning_rate * m_gradient
    b = b - learning_rate * b_gradient
 
    if epoch % 200 == 0:
        cost = np.mean(error ** 2)   # MSE - same cost function from earlier
        print(f"Epoch {epoch}: m={m:.2f}, b={b:.2f}, cost={cost:.2f}")
 
print(f"\nFinal learned line: marks = {m:.2f} * hours + {b:.2f}")
```

### Visualize the fitted line

```python
plt.scatter(hours, marks, color="red", label="Actual data")
plt.plot(hours, m * hours + b, color="blue", label="Learned line")
plt.xlabel("Hours studied")
plt.ylabel("Marks scored")
plt.legend()
plt.title("From-scratch Linear Regression")
plt.show()
```
