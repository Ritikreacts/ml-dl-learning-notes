```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("blastchar/telco-customer-churn")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os 


csv_file = os.path.join(path,'WA_Fn-UseC_-Telco-Customer-Churn.csv')
df = pd.read_csv(csv_file)
df['Churn'].value_counts()
```

```python
pip install scipy
```

```python
from scipy import stats

churned_tenure = df[df["Churn"] == "Yes"]["tenure"]
non_churned_tenure = df[df["Churn"] == "No"]["tenure"]

t_stat, p_value = stats.ttest_ind(non_churned_tenure, churned_tenure)

print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.5f}")
print(f"Non-churned avg tenure: {non_churned_tenure.mean():.2f}")
print(f"Churned avg tenure: {churned_tenure.mean():.2f}")
```

### Task 18 — Interpretation of the T-Test Result

**What we can claim:**
- There is a statistically significant difference in average tenure between churned and non-churned customers (t = 31.58, p < 0.00001)
- Non-churned customers have meaningfully longer tenure, this isn't random noise, the large sample size (7043) and near-zero p-value make that clear
- Tenure looks like a strong candidate feature for predicting churn

**What we cannot claim:**
- That low tenure *causes* churn, this is correlation, not causation (both could be driven by something else, e.g. contract type — newer customers may disproportionately be on month-to-month plans, which independently drives churn)
- That every churned customer has low tenure, this is a difference in *averages*, individual customers vary a lot
- That tenure alone can predict churn well, a t-test only compares two groups' means; it says nothing about how good a full predictive model would be
